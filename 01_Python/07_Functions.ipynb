### Scope of a variable

```python
a = 1
def sample():
  a += 1
  print(a)

print(a)
sample()
```

```python
a = 1
def sample():
  a = 3
  a += 1
  print(a)

print(a)
sample()
```

```python
a = [10 , 20 , 30]
def sample():
  a.append(60)
  print(a)
print(a)
sample()
print(a)
```

```python
a = 100
def add():
  global a
  a += 12
  print(a)
```

```python
print(a)
add()
```

### arguments and keyword arguments

```python
def sample(a, b, c):
  result = a + b + c
  return result
```

```python
sample(10, 20, 30)
```

```python
def sample_v1(*args):
  print(type(args))
  print(args)
```

```python
sample_v1(10, 20, 30)
```

```python
def sample_v2(a , b = 100 ,*args):
  print(a , type(a))
  print(b , type(b))
  print(args, type(args))
```

```python
sample_v2(10 , 20 , 30 , 40 , 50 , 60)
```

```python
def sample_v3(a , b = 100 ,*args , value):
  print(a , type(a))
  print(b , type(b))
  print(args, type(args))
  print(value)
```

```python
sample_v3(34 , 67, 53, 4 , 562 , 535)
```

```python
def sample_v4(a , b = 100 ,*args , **kwargs):
  print(a , type(a))
  print(b , type(b))
  print(args, type(args))
  print(kwargs , type(kwargs))
```

```python
sample_v4(34 , 67, 53, 4 , 562 , 535 , d = 100 , e = 200 , f = 300)
```
