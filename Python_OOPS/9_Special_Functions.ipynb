### Special Functions

```python
def add(a , b):
  return a + b
```

```python
add(23 , 23)
```

```python
add_lambda = lambda a , b : a + b
```

```python
add_lambda(23 , 23)
```

```python
square_number = lambda a : a * a
```

```python
square_number(45)
```

### Map Function

```python
lst = [1 , 2, 3, 4, 5]
```

```python
square = list(map(square_number , lst))
```

```python
square
```

```python
names = ["manoj" , "ram" , "sam"]
upper_case_names = list(map(str.upper , names))
```

```python
print(upper_case_names)
```

```python
cube = lambda a : a * a * a
```

```python
numbers = [1 , 2 , 3, 4, 5]
cube_numbers = list(map(cube , numbers))
```

```python
print(cube_numbers)
```

```python
numbers = [1 , 2 , 3, 4 , 5]
result = list(map(lambda a : a * 2 , numbers))
```

```python
print(result)
```

### Filter Function

```python
def is_even(number):
  return number % 2 == 0
```

```python
numbers = [1, 2, 3, 4, 5]
even_numbers = list(filter(is_even , numbers))
```

```python
print(even_numbers)
```

```python
odd_numbers = list(filter(lambda a : a % 2 != 0 , numbers))
```

```python
print(odd_numbers)
```

```python
students = {
  "manoj" : "batch_1",
  "ram" : "batch_2"

}
```

```python
print(students)
```

```python
for i in students.items():
  print(i)
```

```python
def is_batch_1(student):
  return student[1] == "batch_1"
```

```python
batch_1_students = list(filter(is_batch_1 , students.items()))
```

```python
print(batch_1_students)
```

```python
names = ["manoj" , "ram"]
batch = ["batch_1" , "batch_2"]
```

```python
batch_1_students_new = list(filter(is_batch_1, list(zip(names , batch))))
```

```python
print(batch_1_students_new)
```

### List Comprehension

```python
numbers = [i for i in range(1 , 20)]
```

```python
print(numbers)
```

```python
even_numbers = [i for i in range(10) if i % 2 == 0]
```

```python
print(even_numbers)
```

```python
even_numbers_flag = [True if i % 2 == 0 else False for i in numbers]
```

```python
even_numbers_flag
```

### Dictonary Comprehension

```python
squares = {x: x**2 for x in range(1, 6)}
print(squares)
```

```python
original_dict = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
even_dict = {k: v for k, v in original_dict.items() if v % 2 == 0}
print(even_dict)
```

```python
numbers = [1, 2, 3, 4]
classified = {x: ("even" if x % 2 == 0 else "odd") for x in numbers}
print(classified)
```

```python
keys = ['name', 'age', 'role']
values = ['Alice', 30, 'Engineer']

user_profile = {k: v for k, v in zip(keys, values)}
print(user_profile)
```

```python
fahrenheit = {'New York': 32, 'Miami': 68, 'LA': 50}
celsius = {city: round((temp - 32) * 5/9) for city, temp in fahrenheit.items()}
print(celsius)
```
