### numpy chars and strings

#### 1. np.char.add()

```python
a = np.array(['hello ' , 'manoj '])
b = np.array(['world' , 'babu'])
```

```python
print(np.char.add(a , b))
```

```python
print(np.char.multiply(a , 2))
```

```python
print(np.char.upper(b))
```

```python
print(np.char.lower(a))
```

```python
print(np.char.capitalize(a))
```

```python
print(np.char.center(b , 11 , fillchar = "*"))
```

```python
a = np.array(["hello__" , "world__"])
print(a)
```

```python
print(np.char.replace(a, "__" , ""))
```

```python
a = np.array(["data_science" , "machine_learning"])
print(np.char.split(a , sep="_"))
```

```python
print(np.char.find(a, "learn"))
```
