```python
sales_jan = pd.DataFrame([100, 200, 300] , index=["Store_A" , "Store_B" , "Store_C"] , columns=["Sales"])
sales_jan
```

```python
sales_feb = pd.DataFrame([110, 350] , index=["Store_A" , "Store_C"] , columns=["Sales"])
sales_feb
```

```python
common_index = sales_jan.index
print(common_index)
```

```python
sales_feb_aligned = sales_feb.reindex(common_index)
sales_feb_aligned
```

```python
sales_diff = sales_feb_aligned["Sales"] - sales_jan["Sales"]
sales_diff
```

```python
sales_feb_aligned = sales_feb.reindex(common_index , fill_value = 0)
sales_feb_aligned
```

```python
sales_diff = sales_feb_aligned["Sales"] - sales_jan["Sales"]
sales_diff
```

### Iterating in pandas

```python
data = pd.DataFrame({
  "Name" : ["Manoj" , "Ram" , "Sai"],
  "Age" : [26 , 32 , 31],
  "City" : ["Hyderabad" , "Chennai" , "Banglore"]
})
```

```python
data
```

```python
for index, row in data.iterrows():
  print(f"{index} : {row["Name"]} lives in {row["City"]} and of {row["Age"]} years old")
```

```python
for row in data.itertuples():
  print(row)
  print(f"{row.Index} : {row.Name} lives in {row.City} and of {row.Age} years old", end = "\n\n")
```

```python
for col in data.columns:
  print(f"Column : {col}")
  print(data[col].values , end = "\n\n")
```

```python
list(data.columns)
```

```python
data
```

### Conditional Example: Adding Column with custom logic

```python
categories = []
for index, row in data.iterrows():
  if row["Age"] < 30:
    categories.append("Student")
  else:
    categories.append("Employee")
```

```python
print(categories)
```

```python
data["Categories"] = categories
data
```

```python
data["Categories_1"] = data["Age"].apply(lambda age: "Student" if age < 30 else "Employee")
```

```python
data
```

```python
lambda_age = lambda age : "Student" if age < 30 else "Employee"
data["Categories_2"] = data["Age"].apply(lambda_age)
data
```

```python
def category_check(age):
  if age < 30:
    return "Student"
  else:
    return "Employee"
```

```python
data["Categories_3"] = data["Age"].apply(category_check)
data
```

```python
def check_is_hyd_and_young(row):
  city = row["City"]
  age = row["Age"]
  if city == "Hyderabad":
    if age < 30:
      return True
    else:
      return False
  return False
```

```python
data["is_hyd_and_young"] = data.apply(check_is_hyd_and_young, axis = 1)
```
