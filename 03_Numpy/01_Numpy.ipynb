### np.arange(start, stop, step)

### np.linspace(start , stop , num = x)

```python
np_linspace_1 = np.linspace(1 , 100 , 4)
print(np_linspace_1)
print(type(np_linspace_1))
```

```python
np_linspace_2 = np.linspace(1 , 10 , 10)
print(np_linspace_2)
print(type(np_linspace_2))
```

```python
for num in np.linspace(1 , 10 , 10):
  print(num)
  print(type(num))
  print(type(int(num)))
  print("=================================")
```
