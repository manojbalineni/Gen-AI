```python
import os

print("Try except block started")

try:
  file_name = "demo_files/temp_06.txt"
  with open(file_name) as f:
    contents = f.read()
    print(contents)
except FileNotFoundError:
  print("File error. Unable to find the file")
  print("Current Locstion : " , os.getcwd())
  print("File Name is : " , file_name)
print("try except block completed")
```

```python
number = 0
try:
  result = 100 / number
except FileNotFoundError:
  print("Handling File Not found exception")

except TypeError:
  print("Handling Type Error")


except Exception as e:
  print("Exception occured : " , e)
```
