### Example

```python
import time
```

```python
def my_decorator(func):
  def wrapper():
    t1 = time.perf_counter()
    func()
    t2 = time.perf_counter()
    total_time_taken = t2 - t1
    print("Time taken : " , total_time_taken)
  return wrapper
```

```python
def loop_and_multiply():
  total = 1
  for i in range(1, 10):
    total += (total * i)
    time.sleep(1)
  print(total)
```

```python
loop_and_multiply = my_decorator(loop_and_multiply)
```

```python
loop_and_multiply()
```

```python
@my_decorator
def loop_and_sum():
  total = 0
  for i in range(1 , 10):
    total = total + i
    time.sleep(1)
  print(total)
```

```python
loop_and_sum()
```

### Example

```python
def exception_decorator(func):
  def wrapper():
    try:
      func()
    except Exception as ex:
      print("Error : " , ex)
  return wrapper
      
```

```python
@exception_decorator
def add_numbers():
  a = input("Enter Number :")
  print(a + 100)
```

```python
add_numbers()
```

### Decorator final Concept

```python
def sample_decorator(func):
  def wrapper(a, b):
    try:
      func(a , b)
    except Exception as ex:
      print("Error : " , ex)
  return wrapper
```

```python
def add_v1(a , b):
  c = a + b
  print(c)
```

```python
add_v1 = sample_decorator(add_v1)
```

```python
add_v1()
```

```python
add_v1(10 , 20)
```

```python
def sample_decorator_v1(func):
  def wrapper(*args, **kwargs):
    try:
      func(*args, **kwargs)
    except Exception as ex:
      print("Error : " , ex)
  return wrapper
```

```python
def add_v2(a , b , c):
  d = a + b + c
  print(d)
```

```python
add_v2(10 , 20 , 80)
```

```python
@sample_decorator_v1
def add_v3(*args , **kwargs):
  sum = 0
  for i in args:
    sum += i
  for key , value in kwargs.items():
    sum += value
  print(sum)
```

```python
add_v3(1 , 2, 3 , 4 , b = 100 , c = 200)
```

```python
def sample_decorator_v2(func):
  def wrapper(*args, **kwargs):
    try:
      result = func(*args, **kwargs)
    except Exception as ex:
      print("Error : " , ex)
    return result
  return wrapper
```

```python
@sample_decorator_v2
def add_v4(*args , **kwargs):
  sum = 0
  for i in args:
    sum += i
  for key , value in kwargs.items():
    sum += value
  return sum
```

```python
result = add_v4(1 , 2, 3 , 4 , b = 100 , c = 200)
```

```python
print(result)
```
