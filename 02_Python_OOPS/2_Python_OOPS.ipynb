#### Encapsulation

```python
class BankAccount:
  def __init__(self, name , balance):
    self.name = name  #public
    self._account_type = "Savings Account"  #protected
    self.__balance = balance #private

  def get_balance(self):
    print("Balance in the account is " , self.__balance)

  def add_and_set_balance(self , depoist):
    self.__balance += depoist
    print("Balance in the account is " , self.__balance)
```

```python
bank_account = BankAccount("Manoj" , 100000)
```

```python
bank_account._account_type
```

```python
bank_account.__balance
```

```python
bank_account.get_balance()
```

### Inheritance

Inhertitance allows a class to reuse properties and methods of another class

#### Single Inheritance

```python
class Device:
  def power_on(self):
    print("Power on the device")
```

```python
class TV(Device):
  def play_channel(self):
    print("Playing the channel")
```

```python
tv = TV()
tv.power_on()
tv.play_channel()
```

#### Multilevel Inheritance

```python


class Device:
  def power_on(self):
    print("Device is getting powered on")

class TV(Device):
  def play_channel(self):
    print("Playing Channel")

class SmartBulb(TV):
  def light_on(self):
    print("Lighting ON the smart bulb")
```

```python
smartBulb = SmartBulb()
smartBulb.power_on()
smartBulb.play_channel()
smartBulb.light_on()
```

#### Multiple Inheritance

```python
class Camera:
  def click_photo(self):
    print("Clicking a photo")

	
```

```python
class WhatsApp:
  def send_message(self):
    print("Sending a message")
```

```python
class MobilePhone(Camera , WhatsApp):
  def make_call(self):
    print("Making a phone call")
```

```python
mobile_phone = MobilePhone()
mobile_phone.make_call()
mobile_phone.send_message()
mobile_phone.click_photo()
```

### Method overriding

```python
class Animal:
  def make_sound(self):
    print("Animal is making sound")
```

```python
class Dog(Animal):
  def make_sound(self):
    super().make_sound()  # Animal().speak()
    print("Dog is making sound")
```

```python
dog = Dog()
dog.make_sound()
```

```python
class A:
  def __init__(self):
      print("I am inside class A")

  def speak(self):
     print("Parent is speaking")
    
```

```python
class B(A):
  def __init__(self):
    super().__init__()
    super().speak()

  def make_sound(self):
    print("Child is speaking")
```

```python
obj = B()
```

```python
obj.make_sound()
```

### Example


```python
class MotherBoard:
  def __init__(self, ram, cpu_speed , cpu_type):
    self.ram = ram
    self.cpu_speed = cpu_speed
    self.cpu_type = cpu_type

  def start(self):
    print("Power flow started")

  def stop(self):
    print("Power flow stopped")
```

```python
class Computer(MotherBoard):
  def __init__(self, ram , cpu_speed , cpu_type):
    super().__init__(ram , cpu_speed , cpu_type)
    self.os_type = self.__check_and_os_type()

  def __check_and_os_type(self):  #private method
    if self.cpu_type.lower() == "mac-chip":
      return "MAC"
    else:
      return "Other"

  def check_my_os(self):  # public method
    print("User has  : " , self.os_type )
    
	
```

```python
computer = Computer("12gb", "24.86ghz" , "mac-chip")
```

```python
computer.check_my_os()
```

```python
computer.__check_and_os_type()
```

```python
computer.start()
```

```python
computer.stop()
```

```python
computer.ram
```

### Abstraction(Contract)

```python
from abc import ABC , abstractmethod
```

```python
class Payment(ABC): #requriment no - 1

  @abstractmethod   # decorator
  def pay(self , amount , currency):  #argument matching is not mandatory
    pass

  @abstractmethod
  def refund(self):
    pass


# Contract called Payment has 2 methods in the contract which all the banks needs to implement
# First is pay
# Second is refund
  
```

```python
class ManojBank(Payment):

  def pay(self):
    print("This method contains logic to pay")

  def refund(self):
    print("This method contains logic to refund")

  def new_user(self):
    print("New user registration")
```

```python
manoj_bank = ManojBank()
```

```python
manoj_bank.pay()
```

```python
manoj_bank.refund()
```

```python
manoj_bank.new_user()
```
