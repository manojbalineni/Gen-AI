```python
import numpy as np
```

```python
data = [10, 20, 30, 40, 100, 48, 1000]
```

```python
print(f"Mean of data : {np.mean(data)}")
```

```python
print(f"Median of data : {np.median(data)}")
```

```python
print(f"Variance of data : {np.var(data)}")  # population variance -> denominator -> n
```

```python
print(f"Standard deviation of data : {np.std(data)}")
```

```python
import statistics
```

```python
print(f"Mode of data : {statistics.mode(data)}")
```

```python
print(f"Variance of data : {np.var(data , ddof = 1)}")  # sample variance -> denominator -> n - 1 -> n - ddof

# ddof means delta degress of freedom
```
