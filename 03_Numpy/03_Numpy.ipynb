### ravel()

```python
import numpy as np
```

```python
array = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
print(array)
```

```python
array_ravel = array.ravel()
print(array_ravel)
```

```python
print(id(array))
print(id(array.ravel()))
print(id(array_ravel))
```

```python
print(f"Data address of 'a' :{array_ravel.__array_interface__['data'][0]}")
```

```python
print(f"Data address of 'a' :{array.__array_interface__['data'][0]}")
```

```python
print(array)
```

```python
array[0][0] = 11
print(array)
```

```python
print(array.ravel())
```

```python
print(array.ravel().shape)
```

### flatten()

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a)
```

```python
print(f"Address of 'a' : {a.__array_interface__['data'][0]}")
```

```python
a_flatten = a.flatten()
print(a_flatten)
```

```python
print(f"Address of 'a_flatten' : {a_flatten.__array_interface__['data'][0]}")
```

### squeeze()

```python
a = np.array([[[1],[2],[3]]])
print(a)
print(a.shape)
```

### Boolean Indexing

```python
a = np.array([5, 10, 15, 20, 25])
print(a)
```

```python
print([a > 15])
```

```python
print(a[a > 15])
```

### Fancy Indexing

```python
a = np.array([100, 200, 300, 400, 500])
print(a)
```

```python
indices = [0, 2, 4]
print(a[indices])
```

### Modifying with indexing

```python
arr = np.array([1, 2, 3, 4])
arr[2] = 100
print(arr)
```

```python
arr[arr > 2] = 2000
print(arr)
```

### Indexing with np.where()

```python
arr = np.array([10, 20, 30, 40, 50, 60])
print(arr)
```

```python
idx = np.where(arr > 20)
print(idx)
```
