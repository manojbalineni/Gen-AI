```python
import sqlite3
```

```python
conn = sqlite3.connect("student_1.db")
cursor = conn.cursor()
```

```python
cursor.execute("PRAGMA foriegn_keys = ON;")
```

```python
cursor.execute(
  '''
CREATE TABLE sample_student (
  id INT PRIMARY KEY,
  name TEXT NOT NULL,
  batch_id INT NOT NULL
)

'''
)
```

```python
student_values = (1012, "Manoj" , 2)
```

```python
cursor.execute('''

INSERT INTO sample_student (id , name , batch_id)
VALUES (? , ?, ?)

''' , student_values)
```

```python
conn.commit()
```

```python
conn.close()
```
