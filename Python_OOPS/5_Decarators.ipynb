### Decorators

```python
def add():
  print("Add function")
```

```python
add()
```

```python
add
```

```python
print(id(add))
```

```python
x = add
```

```python
print(id(x))
```

```python
x()
```

```python
def greet():
  print("Hello I am greeting")
```

```python
def sample(a):
  a()
```

```python
sample(1)
```

```python
sample(greet)
```

```python
print(greet)
print(id(greet))
```
