```python
class BankAccount:

	def __init__(self , name , balance):
		self.name = name
		self.balance = balance

	def withdraw(self , amount):
		if amount > self.balance:
			print("Insufficient Balance")
		else:
			self.balance -= amount
		print(self.name, "withdraw : " , amount , ". New Balance : " , self.balance)

```

```python
user_1 = BankAccount("Manoj" , 1000)
user_1.withdraw(100)
```

```python
print(type(user_1))
```

```python
user_1.withdraw(453)
```

### Creating own classes

```python
class Robot:
	# init or constructor is a special function that runs when you create a object from class (blue print)
	# init is optional
	# used to intialize or declare varaibles
	def __init__(self):
		print("Hello, Welcome to new Robot")

	def walk(self):
		print("Robot is walking 10 mts")

	def sing(self):
		print("Robot is Sining")
		
```

```python
a = Robot()
a.walk()
a.sing()
```

```python
print(type(a))
```

```python
print(id(a))
```

```python
class Robot:
	def __init__(self):
		print("Hello, Welcome to new Robot")
		self.value = 1000

	def walk(self):
		print("Robot is walking 10 mts")
		print(self.value)

	def sing(self):
		print("Robot is Sining")
		print(self.value)
		
```

```python
a = Robot()
a.value
```

```python
a.walk()
```

```python
a.sing()
```

```python
a.value = 2000
```

```python
a.sing()
```

```python
class Robot:
	def __init__(self , customer_name):
		print("Hi , Welcome to the Robot")
		self.name = "Chitti"
		self.ram_in_gb = 10
		self.version = "0.0.1"
		self.customer_name = customer_name


	def config(self):
		print("Version : " , self.version)
		print("Name : " , self.name)
		print("Ram : " , self.ram_in_gb)

	def customer_details(self):
		print("Customer Details : " , self.customer_name)

  
```

```python
a = Robot("Manoj")
```

```python
a.config()
```

```python
a.customer_details()
```

```python
r2 = Robot("Ram")
```

```python
r2.config()
```

```python
r2.customer_details()
```
