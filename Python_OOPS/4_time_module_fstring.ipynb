### time module

```python
import time
```

```python
t1 = time.time()
```

```python
print(t1)
```

```python
t2 = time.time()
```

```python
t2 - t1
```

```python
t1_perf = time.perf_counter()
print(t1_perf)
```

```python
t2_perf = time.perf_counter()
print(t2_perf)
```

```python
t2_perf - t1_perf
```

```python
def add(a , b):
  return a + b
```

```python
add(56 , 67)
```

```python
t1 = time.time()
add(1 , 1)
t2 = time.time()
print(t2 - t1)
```

```python
t1_perf = time.perf_counter()
add(1 , 1)
t2_perf = time.perf_counter()
print(t2_perf - t1_perf)
```

```python
# if I want my python program to wait 

print("Hello World")
time.sleep(3)  # Wait for 3 seconds
print("Manoj")
```

```python
t3 = time.time()
print(t3 - t1)
```

```python
hours , remainder = divmod(t3 - t1 , 3600)
minutes , seconds = divmod(remainder , 60)
```

```python
print(hours)
print(minutes)
print(seconds)
```

```python
print("Time taken H : " , int(hours) , "M : " , int(minutes) , "S : " , seconds)
```
