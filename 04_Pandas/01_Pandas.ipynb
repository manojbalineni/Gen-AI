### Series Creation

#### From List

```python
data = [1, 2, 3, 4, 5]
s1 = pd.Series(data)
print(s1)
```

```python
s2 = pd.Series(data, index = ['a', 'b' , 'c' , 'd' , 'e'] , name = "Scores")
print(s2)
```

#### From dictionary

```python
marks = {'manoj' : 23 , 'ram' : 24 , 'sam' :25}
s3 = pd.Series(marks , name = "Marks of DS")
print(s3)
```

#### From Numpy Array

```python
import numpy as np
```

```python
array = np.array([10, 20, 30, 40, 50])
s4 = pd.Series(array, index = ['a' , 'b', 'c', 'd' , 'e'])
print(s4)
```
