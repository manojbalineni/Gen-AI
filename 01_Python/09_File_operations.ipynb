```python
with open("demo_files/temp_05.txt" , "w+") as f:
  first_index = f.tell()
  print(first_index)

  f.write("Manoj Balineni is learning data science")
  end_index = f.tell()
  print(end_index)

  middle_index = end_index // 2  + 1
  print(middle_index)

  f.seek(middle_index)
  f.write("**************")
  print(f.tell())
```

```python
# Add data to the middle of the file, without overwritting the existing file

with open("demo_files/temp_05.txt" , "w+") as f:
  first_index = f.tell()
  print(first_index)

  f.write("File Opeartions in python")
  last_index = f.tell()
  print(last_index)

  middle_index = (last_index) // 2 + 1
  print(middle_index)

  f.seek(middle_index)
  middle_last_content = f.read()
  print(middle_last_content)
  
  f.seek(middle_index)
  f.write("********")
  f.write(middle_last_content)

```
