```python
import numpy as np
```

```python
np.random.seed(88)
```

```python
population = np.random.normal(loc = 50 , scale = 10 , size = 100000)

# normal distribution
# loc means the mean should be 50
# scale means the SD should be 10
```

```python
population.shape
```

```python
population[:10]
```

```python
population_varaince = np.var(population) # This uses 1/n by default
print(f"True Variance of the population : {population_varaince}")
```

```python
n_samples = 1000
sample_size = 5

var_using_n = []
var_using_n_minus_1 = []
```

```python
for _ in range(n_samples):
  sample = np.random.choice(population , size = sample_size , replace = False)
  var_n = np.var(sample)
  var_n_minus_1 = np.var(sample , ddof = 1)
  var_using_n.append(var_n)
  var_using_n_minus_1.append(var_n_minus_1)
```

```python
avg_var_n = np.mean(var_using_n)
print(avg_var_n)
```

```python
avg_var_n_minus_1 = np.mean(var_using_n_minus_1)
print(avg_var_n_minus_1)
```

```python
import numpy as np
```

```python
def detect_outliers(data , k = 1.5):
  data.sort()

  arr = np.array(data , dtype = float)
  q1 = np.percentile(arr , 25 , method="linear")
  q3 = np.percentile(arr , 75 , method="linear")
  iqr = q3 - q1
  lower_fence = q1 - k * iqr
  upper_fence = q3 + k * iqr

  mask = (arr >= lower_fence) & (arr <= upper_fence) 

  #to extract outliers
  outliers_mask = ~mask

  return {
    "Q1" : float(q1),
    "Q3" : float(q3),
    "IQR" : float(iqr),
    "Lower Fence" : float(lower_fence),
    "Upper Fence" : float(upper_fence),
    "Clean Data" : arr[mask].tolist(),
    "Outliers" : arr[outliers_mask].tolist()

  }

```

```python
student_scores = [10, 12, 45, 34, 20, 33, 35, 40, 55, 60, 61, 44, 48, 53, 90, 98]
```

```python
detect_outliers(student_scores)
```
