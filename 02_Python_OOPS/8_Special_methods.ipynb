```python
class Wallet:
  def __init__(self,cash):
      self.cash = cash

  def __add__(self, other):
      return Wallet(self.cash + other.cash)

  def __sub__(self, other):
     return Wallet(self.cash - other.cash)

  def __eq__(self, value):
     return self.cash == value.cash
```

```python
w1 = Wallet(40)
w2 = Wallet(30)

total_sum = w1 + w2
print(total_sum.cash)
```

```python
w1 = Wallet(40)
w2 = Wallet(30)

total_diff = w1 - w2
print(total_diff.cash)
```

```python
w1 = Wallet(40)
w2 = Wallet(40)

print(w1 == w2)
```

```python
class ATM:
  def __init__(self):
    self.transactions = []

  def __len__(self):
    return len(self.transactions)


  def do_transactions(self , value):
    self.transactions.append(value)
    return f"Done Transaction of {value}"

  def __getitem__(self, key):
    return self.transactions[key]
```

```python
atm = ATM()
```

```python
atm.do_transactions(30)
```

```python
atm.do_transactions(50)
```

### Property Decorator

```python
class Account:
  def __init__(self , balance):
    self.__balance = balance

  #Getter
  @property
  def balance(self):
    print("In getter method")
    return self.__balance

  @balance.setter
  def balance(self,value):
    if value < 0:
      print("Security Alert !! Balance cannnot be negative")
    elif value > 1000000:
      print("Security Alert !! Larger amount needs approval")
    else:
      print("In setter Method")
      self.__balance = value

  @balance.deleter
  def balance(self):
    print("We cannot delete the balance")
```

```python
account = Account(2000)
print(account.balance)
```

```python
account = Account(100)
print(account.balance)
account.balance = 2000
print(account.balance)
```
