### Mathematical Operations

```python
import numpy as np
```

```python
list_a = [1, 2, 3, 4]
list_b = [5, 6, 7, 8]

print(list_a + list_b)
```

```python
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])
print(a + b)
```

```python
print(a * b)
```

```python
print(a - b)
```

```python
print(a / b)
```
