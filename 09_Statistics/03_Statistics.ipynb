### Correlation using titanic dataset

```python
import seaborn as sns
import pandas as pd
```

```python
# https://github.com/datasciencedojo/datasets
titanic_dataset_github_url = "https://raw.githubusercontent.com/datasciencedojo/datasets/refs/heads/master/titanic.csv"
```

```python
dataset = pd.read_csv(titanic_dataset_github_url)
```

```python
dataset.head()
```

```python
numeric_df = dataset.select_dtypes(include=["int64", "float64"])
```

```python
print(numeric_df.columns)
```

```python
corr_matrix = numeric_df.corr()
```

```python
corr_matrix
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(corr_matrix , annot=True , cmap="coolwarm" , fmt = ".2f" , linewidths=0.5)
plt.title("Correlation heat map for titanic dataset")
plt.show()
```

### Probability Density Function

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
bins = [150, 155, 160, 165, 170, 175, 180]
pdf_values = [0.01, 0 , 0.03, 0.06, 0.06, 0.04]
```

```python
plt.bar(bins[:-1] , pdf_values , width=5 , edgecolor='black' , alpha = 0.7)
plt.title("PDF of heights")
plt.xlabel("Height(cm)")
plt.ylabel("Probability Density")
plt.show()
```
