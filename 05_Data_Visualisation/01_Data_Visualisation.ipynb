### Cleaning and Feature Engineering

```python
df.info()
```

```python
# removing the rows where customer ID is null

df = df.dropna(subset=["Customer ID"])
```

```python
df.info()
```

```python
df.loc[100:200]
```

```python
# removing the cancelled invoices

df = df[~df["Invoice"].astype(str).str.startswith("C")]
```

```python
df.shape
```

```python
824364-805620
```

```python
df.loc[100:200]
```

```python
# Removing the zero/negative quantities

df = df[df["Quantity"] > 0]
```

```python
df.info()
```

```python
# Removing the zero/negative price

df = df[df["Price"] > 0]
```

```python
df.info()
```

```python
df.shape
```

```python
# Creating a revenue column

df['Revenue'] = df['Quantity'] * df['Price']
```

```python
df.head(3)
```

```python
df.info()
```

```python
df["Month"] = df["InvoiceDate"].dt.to_period("M").astype(str)
```

### Subplots : Multiple plots in one figure

```python
df.head()
```

```python
monthly_quantity = df.groupby('Month')['Quantity'].sum().sort_index()
```

```python
monthly_quantity
```

```python
monthly_revenue  = df.groupby('Month')['Revenue'].sum().sort_index()
```

```python
monthly_revenue
```

```python
fig, axs = plt.subplots(nrows = 2 , ncols = 1 , figsize = (10, 8) , sharex=True)

# Plot Monthly revenue
axs[0].plot(monthly_revenue.index , monthly_revenue.values , marker = 'o' , color = 'blue' , label = 'revenue')
axs[0].set_xlabel("Revenue")
axs[0].set_title("Monthly Revenue")
axs[0].set_ylabel("Month")
axs[0].legend()
axs[0].grid(True)



# Plot monthly quantity
axs[1].plot(monthly_quantity.index, monthly_quantity.values , marker = 'o' , color = 'green' , label = 'quantity')
axs[1].set_xlabel("Quantity")
axs[1].set_title("Monthly Quantity")
axs[1].set_ylabel("Month")
axs[1].grid(True)
axs[1].legend()


plt.title("Monthly revenue and quantity")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
fig, axs = plt.subplots(nrows = 2 , ncols = 1 , figsize = (10, 8) , sharex=True)

# Plot Monthly revenue
axs[0].plot(monthly_revenue.index , monthly_revenue.values , marker = 'o' , color = 'blue' , label = 'revenue')
axs[0].plot(monthly_quantity.index, monthly_quantity.values , marker = 'o' , color = 'green' , label = 'quantity')
axs[0].set_xlabel("Revenue")
axs[0].set_title("Monthly Revenue")
axs[0].set_ylabel("Month")
axs[0].legend()
axs[0].grid(True)



# Plot monthly quantity
axs[1].plot(monthly_quantity.index, monthly_quantity.values , marker = 'o' , color = 'green' , label = 'quantity')
axs[1].set_xlabel("Quantity")
axs[1].set_title("Monthly Quantity")
axs[1].set_ylabel("Month")
axs[1].grid(True)
axs[1].legend()


plt.title("Monthly revenue and quantity")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
