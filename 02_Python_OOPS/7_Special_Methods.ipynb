### Special Methods(OOPS)

```python
class Car:
	def __init__(self, brand , year):
		self.brand = brand
		self.year = year
	
	def __str__(self):
		return f"Brand Name is {self.brand} and from year {self.year}"

	def __repr__(self):
		return f"Brand Name is {self.brand}. This object contains 2 varaibles"
```

```python
car = Car("Toyota" , 2026)
```

```python
str(car)
```

```python
repr(car)
```

```python
car
```

```python
print(f"{car!s}")
```

```python
print(f"{car!r}")
```
