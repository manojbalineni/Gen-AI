### OOPS - Polymorphism

```python
class TV:
  def start_tv(self):
    print("TV is starting")

class Microwave:
  def start_microwave(self):
    print("Microwave is starting")

class Fan:
  def start_fan(self):
    print("Fan is starting")
```

```python
tv = TV()
tv.start_tv()

microwave = Microwave()
microwave.start_microwave()

fan = Fan()
fan.start_fan()
```

```python
class TV:
  def start(self):
    print("TV is starting")

class Microwave:
  def start(self):
    print("Microwave is starting")

class Fan:
  def start(self):
    print("Fan is starting")

```

```python
electical_appliances = [TV() , Microwave() , Fan()]
for i in electical_appliances:
  i.start()
```

```python
from abc import ABC , abstractmethod
```

```python
class ElecticalAppliance(ABC):

  @abstractmethod
  def start(self):
    pass

  
```

```python
class TV(ElecticalAppliance):
  def start(self):
    print("TV is starting")

class Microwave(ElecticalAppliance):
  def start(self):
    print("Microwave is starting")

class Fan(ElecticalAppliance):
  def start(self):
    print("Fan is starting")

class AirFyer(ElecticalAppliance):
  def start(self):
    print("Airfyer is starting")
```

```python
electical_appliances = [TV() , Microwave() , AirFyer() , Fan()]
```

```python
for appliance in electical_appliances:
  appliance.start()
```
