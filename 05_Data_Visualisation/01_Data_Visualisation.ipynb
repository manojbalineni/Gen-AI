### Cleaning and Feature Engineering

```python
df.info()
```

```python
# removing the rows where customer ID is null

df = df.dropna(subset=["Customer ID"])
```

```python
df.info()
```

```python
df.loc[100:200]
```

```python
# removing the cancelled invoices

df = df[~df["Invoice"].astype(str).str.startswith("C")]
```

```python
df.shape
```

```python
824364-805620
```

```python
df.loc[100:200]
```

```python
# Removing the zero/negative quantities

df = df[df["Quantity"] > 0]
```

```python
df.info()
```

```python
# Removing the zero/negative price

df = df[df["Price"] > 0]
```

```python
df.info()
```

```python
df.shape
```

```python
# Creating a revenue column

df['Revenue'] = df['Quantity'] * df['Price']
```

```python
df.head(3)
```

```python
df.info()
```

```python
df["Month"] = df["InvoiceDate"].dt.to_period("M").astype(str)
```

```python
df.head()
```

```python
df.info()
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt
```

```python
monthly_revenue = df.groupby('Month')['Revenue'].sum().sort_index()
monthly_revenue.head()
```
