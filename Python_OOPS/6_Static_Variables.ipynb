### Static Variables

```python
print("Hello World")
```

```python
class User:
  course_name = "Data Science"

  def __init__(self, batch_no):
    self.batch_number = batch_no
	
```

```python
obj1 = User(2)
obj2 = User(3)
```

```python
print(obj1.course_name)
print(obj1.batch_number)
```

```python
print(obj2.course_name)
print(obj2.batch_number)
```

```python
User.course_name = "Gen AI"
```

```python
print(obj1.course_name)
print(obj1.batch_number)
```

```python
print(obj2.course_name)
print(obj2.batch_number)
```

### Static Methods

```python
class MathUtilis:
  
  @staticmethod
  def add(a , b):
    return a + b

  @staticmethod
  def multiply(a , b):
    return a * b
```

```python
MathUtilis.add(20 ,30)
```

```python
MathUtilis.multiply(10 ,100)
```
